```python
#task 1_1
def task1_1(input_value):
    str_value = str(input_value)
    if not str_value.isdigit() or len(str_value) != 14:
        return -1
    total_sum = 0
    for right_pos in range(1,15):
        digit = int(str_value[-right_pos])

        if right_pos % 2 != 0: #means it is an odd number
            doubled = digit * 2
            if doubled >= 10:
                doubled = (doubled // 10) + (doubled % 10)
            total_sum = total_sum + doubled
        else:
            total_sum = total_sum + digit
    remainder = total_sum % 10
    if remainder == 0: #means the there is no remainder
        check_digit = 0
    else:
        check_digit = 10 - remainder

    return check_digit
```

```python
#task 1_2
#create 4 test cases
print(task1_1("14576567654930") == 0) #string containing valid integer
print(task1_1(14576567654930) == 0) #valid integer
print(task1_1("1457656765493A") == -1) #character that are not numbers
print(task1_1(123456) == -1) #incorrect length
```

```python
#task 1_3
def task1_3(input_value):
    str_value = str(input_value)
    if not str_value.isdigit() or len(str_value) != 15:
        return False
    fourteen_digits = str_value[:-1]
    original_check_digit = int(str_value[-1])
    calculated_check_digit = task1_1(fourteen_digits)

    return calculated_check_digit == original_check_digit

```
